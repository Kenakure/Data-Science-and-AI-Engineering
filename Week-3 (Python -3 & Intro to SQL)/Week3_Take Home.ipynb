```python
#1. **JSON File Handling**  
#Write a program that stores a list of student records (name, age, grade) into a JSON file, and then reads the file back into a Python object. 

import json

# Create student records
students = [
    {"name": "Ken Akure", "age": 40, "grade": "A"},
    {"name": "Bolaji", "age": 22, "grade": "B"},
    {"name": "Chioma", "age": 28, "grade": "C"}
]

# write/save to JSON file
with open("students.json", "w") as f:
    json.dump(students, f, indent=4)

print("Data saved to students.json")

# Read it back
with open("students.json", "r") as f:
    loaded_students = json.load(f)

print("\nData loaded from file:")
for student in loaded_students:
    print(f"  {student['name']}, Age: {student['age']}, Grade: {student['grade']}")
```

```python
#2. **Math Quiz**  
   #Build a program that asks the user 5 random math questions (addition, subtraction, multiplication, division) and keeps score of correct answers. 

import random

score = 0
total_questions = 5
operations = ["+", "-", "*", "/"]

print("Math Quiz!")
print("Answer each question. For division, round to 2 decimal places.\n")

for i in range(1, total_questions + 1):
    num1 = random.randint(1, 20)
    num2 = random.randint(1, 20)
    op = random.choice(operations)

    if op == "+":
        correct = num1 + num2
    elif op == "-":
        correct = num1 - num2
    elif op == "*":
        correct = num1 * num2
    elif op == "/":
        correct = round(num1 / num2, 2)

    try:
        user_answer = float(input(f"Q{i}: What is {num1} {op} {num2}? "))
        if user_answer == correct:
            print("Correct!\n")
            score += 1
        else:
            print(f"Wrong! The answer was {correct}\n")
    except ValueError:
        print(f"Invalid input! The answer was {correct}\n")

print("Quiz complete!")
print(f"Your score: {score}/{total_questions}") 

```

```python
#3. **Error Handling System**  
   #Create a program that asks the user for a number and divides 100 by that number. Use `try/except` to handle invalid input and division by zero. 

while True:
    try:
        num = float(input("Enter a number: "))
        result = 100 / num
        print(f"100 / {num} = {result:.2f} ")
        break
    except ValueError:
        print("That's not a number! Try again.\n")
    except ZeroDivisionError:
        print("Cannot divide by zero! Try again.\n")
```

```python
#4. **Date Calculator**  
   #Write a program that takes a user’s birthday as input and tells them how many days are left until their next birthday.

from datetime import date

#Get user's birth_date.

birth_month = int(input("Enter your birth month (1-12): "))
birth_day = int(input("Enter your birth day (1-31): "))

#get today's date

today = date.today()

#figure out next birthday

next_birthday = date(today.year, birth_month, birth_day)

if next_birthday < today:
    next_birthday = date(today.year + 1, birth_month, birth_day)

#calculate the difference

days_left = (next_birthday - today).days

#If birthday is today.

if days_left == 0:
    print("Happy Birthday!")
else:
    print(f"Your next birthday is on {next_birthday.strftime('%B %d, %Y')}")
    print(f"That's {days_left} days away!")
```

#8. Explain the difference between `import module` and `from module import function`.

import module brings in the entire module. You access everything with the module name as a prefix. E.g import math, then you can access the sqrt function with math.sqrt(64).

from module import function brings in only the specified function from the module. You can access the function directly without the module name prefix. E.g from math import sqrt, then you can access the sqrt function with sqrt(64).

#10. What is a virtual environment in Python, and why is it useful?  

A virtual environment is a separate, isolated Python setup for a specific project. It has its own modules and packages that don't affect anything else your global environment (computer).

Without it, all projects share the same packages. Updating a package for one project could break another. With virtual environments, each project has its own space and they never interfere.

```python
#11. Write a program that calculates the number of days between two given dates.

from datetime import date

date1 = input("Enter the first date (YYYY-MM-DD): ")
date2 = input("Enter the second date (YYYY-MM-DD): ")

date1 = date.strptime(date1, "%Y-%m-%d")
date2 = date.strptime(date2, "%Y-%m-%d")

difference = abs((date2 - date1).days)

print("The number of days between the two dates is:", difference)    
```
