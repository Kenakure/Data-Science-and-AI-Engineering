#8. Explain the difference between `import module` and `from module import function`.

import module brings in the entire module. You access everything with the module name as a prefix. E.g import math, then you can access the sqrt function with math.sqrt(64).

from module import function brings in only the specified function from the module. You can access the function directly without the module name prefix. E.g from math import sqrt, then you can access the sqrt function with sqrt(64).


```python
#9. Write a program that asks the user for two numbers and prints their greatest common divisor (GCD). 

import math

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

gcd = math.gcd(num1, num2)

print("The GCD of", num1, "and", num2, "is:", gcd)
```

#10. What is a virtual environment in Python, and why is it useful?  

A virtual environment is a separate, isolated Python setup for a specific project. It has its own modules and packages that don't affect anything else your global environment (computer).

Without it, all projects share the same packages. Updating a package for one project could break another. With virtual environments, each project has its own space and they never interfere.

```python
#11. Write a program that calculates the number of days between two given dates.

from datetime import date

date1 = input("Enter the first date (YYYY-MM-DD): ")
date2 = input("Enter the second date (YYYY-MM-DD): ")

date1 = date.strptime(date1, "%Y-%m-%d")
date2 = date.strptime(date2, "%Y-%m-%d")

difference = abs((date2 - date1).days)

print("The number of days between the two dates is:", difference)    
```
