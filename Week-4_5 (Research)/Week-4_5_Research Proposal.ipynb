### 1. Background and Problem Statement

#### 1.1 Background

Blockchain technology and cryptocurrencies have grown significantly over the last decade. Ethereum, in particular, has become one of the most widely used blockchain networks, supporting everything from simple cryptocurrency transfers to decentralised finance applications, tokens and smart contracts.

However, increased adoption has also created more opportunities for fraud. Cryptocurrency-related scams, phishing attacks, money laundering and other illicit activities continue to result in significant financial losses. One of the challenges is that blockchain transactions are pseudonymous. Although transactions can be viewed publicly, identifying whether an address or transaction is connected to fraudulent activity is not always straightforward.

Many fraud detection systems still depend heavily on predefined rules. For example, a system may flag transactions above a particular value or transactions involving addresses that have already been blacklisted. These approaches are useful, but they have an obvious weakness: they usually react to behaviour that is already known.
Fraudsters, on the other hand, continuously change their methods.
This is where machine learning becomes useful. Rather than relying entirely on predefined rules, a machine learning model can learn patterns from historical transaction data and use those patterns to identify potentially fraudulent behaviour. This makes it possible to detect relationships and anomalies that may not be immediately obvious through traditional rule-based approaches.

#### 1.2 Problem Statement

Traditional rule-based fraud detection systems are useful for identifying known fraudulent behaviour, but they are less effective when fraud patterns become more complex or change over time.
The challenge becomes even greater in cryptocurrency datasets because fraudulent transactions usually represent only a small portion of the overall data. A model can therefore appear to perform well while still failing to identify many fraudulent transactions.

This research will investigate whether machine learning models can provide a more effective approach to identifying fraudulent activity on the Ethereum blockchain. It will compare several classification algorithms and examine how well they perform when dealing with imbalanced transaction data.

#### 1.3 Why This Matters

Fraud detection is relevant across different parts of the cryptocurrency ecosystem.
Wallet infrastructure providers and regulated custodians need ways to protect the businesses using their platforms. Fintech companies, neobanks and cryptocurrency exchanges also need effective transaction-monitoring systems to support their compliance and Anti-Money Laundering obligations.

Compliance teams need tools that can identify suspicious behaviour beyond addresses that are already known or blacklisted. At the same time, individual cryptocurrency users also benefit from systems that can identify potentially fraudulent addresses and transactions.

The practical challenge is therefore not simply to build a model that performs well academically, but to understand which machine learning approach could realistically contribute to a fraud-detection system used in the real world.

The main objective of this research is to develop and evaluate machine learning models that can identify fraudulent Ethereum transactions using publicly available blockchain data.

4. Do ensemble methods (Random Forest, XGBoost) outperform individual classifiers (Logistic Regression, SVM) for this specific problem?



### 4. Literature Review

#### 4.1 Current State of Research

There's been a lot of academic interest in applying ML to blockchain fraud detection in recent years. I've anchored this research on three studies from well-regarded publishers.

Asiri (2025) built a Graph Convolution Network for detecting illicit Bitcoin transactions. Published in Nature's Scientific Reports, the results were impressive: 98.5% precision. The core insight was that graph-based approaches can pick up on relational patterns between transaction entities, something traditional feature-based models tend to miss. The downside? Graph-based methods are computationally expensive. Running them in real-time, especially in environments where resources are limited, is a real challenge.

A comprehensive survey published in ACM Computing Surveys (2025), titled "Machine Learning for Blockchain Data Analysis: Progress and Opportunities," mapped out what's been done and what hasn't across the whole field. One finding that stood out to me: supervised learning methods, particularly XGBoost and Random Forest, consistently outperform unsupervised approaches when you have labelled data to work with. The survey also flagged class imbalance as a problem that researchers acknowledge but don't always address properly.

Ahmed and Alabi (2024) published a systematic review in IEEE Access focused on federated learning for crypto fraud detection. The privacy angle was interesting. Their argument is that organisations increasingly want to collaborate on fraud detection without sharing their raw transaction data. In their review, XGBoost hit 95% accuracy, while federated learning approaches managed about 91%.

#### 4.2 Research Gap

Despite the progress, there are gaps.

Most studies look at Bitcoin or Ethereum in isolation. Cross-chain analysis is rare. The class imbalance issue keeps showing up as a known problem, but it's handled differently (or not at all) across studies. And perhaps most importantly for this project: very few studies compare multiple ML algorithms on the same dataset using the same preprocessing. That makes it really hard to say definitively whether Random Forest is better than XGBoost, because they were tested under different conditions.

There's also a disconnect between academic results and practical deployment. How these models would actually perform in a production environment, say at a wallet service provider processing thousands of transactions daily, is rarely discussed.
This research directly targets the comparison gap. Same dataset, same preprocessing, same evaluation metrics, four different algorithms. A fair test.


### 5. Methodology

#### 5.1 Research Design

The study will use a quantitative experimental approach.
Four supervised machine learning algorithms will be trained using the same labelled Ethereum fraud dataset. Their performance will then be compared using the same test data and evaluation metrics.

Using the same dataset and preprocessing process is important because it allows the algorithms to be compared under similar conditions. 

#### 5.2 Dataset

The primary dataset for the study will be the Ethereum Fraud Detection Dataset available on Kaggle.
The dataset contains 9,841 Ethereum account records and 51 transaction-related features. Each record is labelled as either legitimate or fraudulent.

Of the 9,841 records, 7,662 are legitimate, while 2,179 are classified as fraudulent. This corresponds to approximately 77.86% legitimate accounts and 22.14% fraudulent accounts.

Some of the available variables include:

- the average value of Ether sent and received;
- total numbers of incoming and outgoing transactions;
- the time between the first and most recent transaction;
- the number of unique addresses an account has interacted with;
- ERC-20 token transaction statistics; and
- gas-related information.

It's not a massive dataset by ML standards, but it's labelled, publicly available, and has been used in published research before, which makes it a solid foundation.

#### 5.3 Data Preprocessing

Before feeding data into any model, several preprocessing steps are needed.
Cleaning comes first. Removing duplicates and dealing with missing values. Depending on how much data is missing in each feature, I'll either impute or drop.

Then feature selection. Not all 51 features will be useful. Some will be highly correlated with each other (redundant), and some won't have much predictive power. I'll use correlation analysis to trim redundancy and feature importance scores from tree-based models to identify what matters most.

Normalisation is next. Algorithms like SVM and Logistic Regression are sensitive to feature scale. If one feature ranges from 0 to 1 and another from 0 to 1,000,000, the model gets confused. StandardScaler brings everything to the same scale.

For class imbalance, I'm using SMOTE (Synthetic Minority Over-sampling Technique) on the training data. But I'm not just applying it blindly. I'll run every model with and without SMOTE so I can actually measure whether it helps, and by how much.

Finally, an 80/20 train-test split with stratification. Stratification means both the training and test sets maintain the same ratio of fraud to legitimate accounts. Without it, I risk getting a test set with very few fraud cases, which would make evaluation unreliable.

#### 5.4 Machine Learning Algorithms

Four algorithms have been selected for the study.

1. Random Forest
Random Forest combines the predictions of multiple decision trees.
It is particularly useful for this study because it performs well with non-linear relationships and can also provide feature-importance scores. These scores will help identify which transaction characteristics contribute most strongly to fraud detection.

2. XGBoost
XGBoost is a gradient-boosting algorithm that has performed strongly across many structured and tabular machine learning problems.
It also provides several methods for controlling overfitting and includes parameters such as scale_pos_weight, which can be useful when working with imbalanced datasets.

3. Logistic Regression
Logistic Regression will act as the simpler baseline model.
Although it is less complex than Random Forest or XGBoost, it has the advantage of being relatively easy to interpret. Its performance will therefore provide a useful reference point when evaluating whether the more complex models produce a meaningful improvement.

4. Support Vector Machine
Support Vector Machine will also be included because it performs well in high-dimensional classification problems.
Using kernel functions, SVM can model non-linear relationships, making it a useful additional comparison for this binary classification task.

#### 5.5 Evaluation Metrics

Accuracy will not be treated as the only measure of success.

This is important because fraud datasets are usually imbalanced. A model could achieve a reasonably high accuracy simply by predicting the majority class most of the time while still missing many fraudulent transactions.

For this reason, several performance metrics will be considered.

1. Accuracy will show the percentage of predictions that are correct overall.

2. Precision will measure how many transactions identified as fraudulent are actually fraudulent. This is important because too many false alerts could make a detection system difficult to use in practice.

3. Recall will measure how many of the genuinely fraudulent records the model successfully identifies. In fraud detection, this is particularly important because a false negative means that potentially fraudulent activity has been missed.

4. F1-score will provide a balance between precision and recall.
AUC-ROC will be used to evaluate how well the model separates the two classes across different decision thresholds.
A confusion matrix will also be used to examine true positives, true negatives, false positives and false negatives directly.

### 9. Limitations

There are several limitations that should be acknowledged from the beginning.

First, the Kaggle dataset is historical and may not represent every type of fraudulent behaviour currently seen on Ethereum.

Second, the study focuses only on Ethereum. The findings therefore may not transfer directly to blockchains that operate differently, particularly Bitcoin and other UTXO-based networks like Doge, Litecoin or Dash.

Third, the models will be tested offline using historical data. A model that performs well under these conditions may still require additional development before it could be used for real-time transaction monitoring.

Finally, the project will work with the features already available in the dataset. More advanced features based on transaction graphs, smart contract interactions or relationships between addresses may improve performance, but these are outside the scope of the current study.



### 10. Ethical Considerations

The project will use publicly available blockchain transaction data and will not involve human participants.

No personally identifiable information will be collected as part of the study. The dataset is publicly available and will be used only for research purposes.

The machine learning models developed during the project are intended for defensive applications such as identifying potentially fraudulent activity and improving transaction-monitoring systems. This research will not be used to de-anonymise blockchain users or identify individuals behind wallet addresses.
