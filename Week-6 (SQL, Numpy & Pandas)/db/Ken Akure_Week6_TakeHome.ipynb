````python
# Combine all cities into one DataFrame

cities = run_query("""
    SELECT 'London'    AS city, GMT AS Date, [Mean TemperatureC], [Max TemperatureC], [Min TemperatureC],
           [Max Humidity], [ Mean Humidity], [Precipitationmm], [ Mean Wind SpeedKm/h]
    FROM week06.london_2014
    UNION ALL SELECT 'Beijing',  Date, [Mean TemperatureC], [Max TemperatureC], [Min TemperatureC],
           [Max Humidity], [ Mean Humidity], [Precipitationmm], [ Mean Wind SpeedKm/h]
    FROM week06.beijing_pek_2014
    UNION ALL SELECT 'Brasilia', Date, [Mean TemperatureC], [Max TemperatureC], [Min TemperatureC],
           [Max Humidity], [ Mean Humidity], [Precipitationmm], [ Mean Wind SpeedKm/h]
    FROM week06.brasilia_bsb_2014
    UNION ALL SELECT 'CapeTown', Date, [Mean TemperatureC], [Max TemperatureC], [Min TemperatureC],
           [Max Humidity], [ Mean Humidity], [Precipitationmm], [ Mean Wind SpeedKm/h]
    FROM week06.capetown_cpt_2014
    UNION ALL SELECT 'Delhi',    Date, [Mean TemperatureC], [Max TemperatureC], [Min TemperatureC],
           [Max Humidity], [ Mean Humidity], [Precipitationmm], [ Mean Wind SpeedKm/h]
    FROM week06.delhi_del_2014
    UNION ALL SELECT 'Moscow',   Date, [Mean TemperatureC], [Max TemperatureC], [Min TemperatureC],
           [Max Humidity], [ Mean Humidity], [Precipitationmm], [ Mean Wind SpeedKm/h]
    FROM week06.moscow_svo_2014
""")

print(f"Combined dataset: {cities.shape}")
cities.head()
````

````python
# Average temperature per city

avg_temp = cities.groupby("city")["Mean TemperatureC"].mean().sort_values(ascending=False)
print("Average Temperature by City (°C):")
print("=" * 40)
for city, temp in avg_temp.items():
    print(f"  {city:12s} {temp:.1f}°C")
````

````python
# Temperature range per city (hottest and coldest days)

temp_stats = cities.groupby("city").agg(
    avg_temp=("Mean TemperatureC", "mean"),
    max_temp=("Max TemperatureC", "max"),
    min_temp=("Min TemperatureC", "min"),
    temp_range=("Mean TemperatureC", lambda x: x.max() - x.min())
).round(1).sort_values("temp_range", ascending=False)

print("Temperature Statistics by City:")
print("=" * 65)
print(temp_stats)
````

````python
# Humidity comparison

avg_humidity = cities.groupby("city")[" Mean Humidity"].mean().sort_values(ascending=False)
print("Average Humidity by City (%):")
print("=" * 40)
for city, hum in avg_humidity.items():
    print(f"  {city:12s} {hum:.1f}%")
````

````python
# Rainfall comparison

total_rain = cities.groupby("city")["Precipitationmm"].sum().sort_values(ascending=False)
rainy_days = cities[cities["Precipitationmm"] > 0].groupby("city").size()

print("Rainfall Summary:")
print("=" * 50)
for city in total_rain.index:
    rain = total_rain[city]
    days = rainy_days.get(city, 0)
    print(f"  {city:12s} Total: {rain:7.1f}mm | Rainy days: {days}")
````

````python
# Extreme weather days

hot_days = cities[cities["Max TemperatureC"] > 35].groupby("city").size().sort_values(ascending=False)
cold_days = cities[cities["Min TemperatureC"] < 0].groupby("city").size().sort_values(ascending=False)

print("Days above 35°C:")
print(hot_days)
print("\nDays below 0°C:")
print(cold_days)
````

````python
# Seasonal patterns (quarterly breakdown)


cities["Date"] = pd.to_datetime(cities["Date"])
cities["quarter"] = cities["Date"].dt.quarter

quarterly = cities.groupby(["city", "quarter"])["Mean TemperatureC"].mean().round(1).unstack()
quarterly.columns = ["Q1 (Jan-Mar)", "Q2 (Apr-Jun)", "Q3 (Jul-Sep)", "Q4 (Oct-Dec)"]

print("Quarterly Average Temperature (°C):")
print("=" * 70)
print(quarterly)
````

### SQL Task 5 · JOIN across two sources
Join WHO TB data (`week06.who_pop_tb_all`) with World Bank population (`week08.wb_pop_2013`) on the country

**Return:** the **top 5** countries: `Country`, `tb_deaths_per_100k`, highest first.
*Revisit: §7.2–7.3.*

````python
SQL_05 = """ SELECT TOP 5
        w.Country,
        ROUND(100000.0 * w.[TB deaths] / NULLIF(p.[SP.POP.TOTL], 0), 1) AS tb_deaths_per_100k
    FROM week06.who_pop_tb_all AS w
    JOIN week08.wb_pop_2013 AS p ON p.country = w.Country
    WHERE w.[TB deaths] IS NOT NULL
      AND p.[SP.POP.TOTL] IS NOT NULL
    ORDER BY tb_deaths_per_100k DESC

"""
answer("s05", SQL_05)
````

### SQL Task 7 · UNION ALL
Stack the **six** city weather tables in `week06` and compute each city's **average `Mean TemperatureC`**
for 2014, **rounded to 1 decimal place**. Use exactly these city labels:
`'Beijing'`, `'Brasilia'`, `'Cape Town'`, `'Delhi'`, `'London'`, `'Moscow'`.

**Return:** `city`, `avg_mean_temp_c`, sorted by `city` (A→Z).
⚠️ `Mean TemperatureC` is stored as a whole number, so watch the integer-division trap. *Revisit: §8.1–8.2.*

````python
SQL_07 = """ WITH cities AS (
        SELECT 'Beijing'   AS city, [Mean TemperatureC] FROM week06.beijing_pek_2014
        UNION ALL SELECT 'Brasilia',  [Mean TemperatureC] FROM week06.brasilia_bsb_2014
        UNION ALL SELECT 'Cape Town', [Mean TemperatureC] FROM week06.capetown_cpt_2014
        UNION ALL SELECT 'Delhi',     [Mean TemperatureC] FROM week06.delhi_del_2014
        UNION ALL SELECT 'London',    [Mean TemperatureC] FROM week06.london_2014
        UNION ALL SELECT 'Moscow',    [Mean TemperatureC] FROM week06.moscow_svo_2014
    )
    SELECT 
        city,
        ROUND(AVG(1.0 * [Mean TemperatureC]), 1) AS avg_mean_temp_c
    FROM cities
    GROUP BY city
    ORDER BY city ASC

"""
answer("s07", SQL_07)
````

### SQL Task 8 · CTE + window ranking (top-1 per group)
For **London**, find the **hottest day of each month** (highest `Max TemperatureC`). If two days tie,
take the **earlier** date.

**Return:** `month_no`, `day` (a real **date**, convert `GMT` with `TRY_CONVERT(date, GMT)`), `max_temp`,
sorted by `month_no`. *Revisit: §10, §11.3.*

````python
SQL_08 = """ WITH ranked AS (
        SELECT 
            MONTH(TRY_CONVERT(date, GMT)) AS month_no,
            TRY_CONVERT(date, GMT) AS day,
            [Max TemperatureC] AS max_temp,
            ROW_NUMBER() OVER (
                PARTITION BY MONTH(TRY_CONVERT(date, GMT))
                ORDER BY [Max TemperatureC] DESC, TRY_CONVERT(date, GMT) ASC
            ) AS rn
        FROM week06.london_2014
    )
    SELECT month_no, day, max_temp
    FROM ranked
    WHERE rn = 1
    ORDER BY month_no

"""
answer("s08", SQL_08)
````

### SQL Task 9 · Running total
From `week09.comtrade_milk_uk_monthly_14`, compute the UK's **total milk exports per month** (exclude `'World'`)
and a **year-to-date running total**.

**Return:** `Period`, `exports_usd`, `ytd_exports_usd`, sorted by `Period`. *Revisit: §11.4–11.5.*

````python
SQL_09 = """ WITH monthly AS (
        SELECT 
            Period,
            SUM([Trade Value (US$)]) AS exports_usd
        FROM week09.comtrade_milk_uk_monthly_14
        WHERE [Trade Flow] = 'Exports'
          AND Partner <> 'World'
        GROUP BY Period
    )
    SELECT 
        Period,
        exports_usd,
        SUM(exports_usd) OVER (ORDER BY Period) AS ytd_exports_usd
    FROM monthly
    ORDER BY Period

"""
answer("s09", SQL_09)
````
