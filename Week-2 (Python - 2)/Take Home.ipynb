- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive!")
            return
        self.balance += amount
        print(f"Deposited ₦{amount:,.2f}. New balance: ₦{self.balance:,.2f}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive!")
            return
        if amount > self.balance:
            print(f"Insufficient funds! Balance: ₦{self.balance:,.2f}")
            return
        self.balance -= amount
        print(f"Withdrew ₦{amount:,.2f}. New balance: ₦{self.balance:,.2f}")

    def __str__(self):
        return f"Account({self.owner}) — Balance: ₦{self.balance:,.2f}"


class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, interest_rate=0.05):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Interest added: ₦{interest:,.2f}. New balance: ₦{self.balance:,.2f}")


# Create a savings account
savings = SavingsAccount("Ken Akure", 100000, 0.08)
print(savings)

# Deposit money
savings.deposit(50000)

# Add interest
savings.add_interest()

# Withdraw money
savings.withdraw(30000)

# Try to overdraw
savings.withdraw(500000)

# Final balance
print(savings)
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    def __init__(self, n):
        self.n = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.n < 1:
            raise StopIteration
        current = self.n
        self.n -= 1
        return current


for num in CountDown(6):
    print(num)
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
# First, create the sample file
with open("sample.txt", "w") as f:
    f.write("Books written by Chinua Achebe \n")
    f.write("1. Things Fall Apart \n")
    f.write("2. Arrow of God \n")
    f.write("3. Ant Hills of the Savanna \n")
    f.write("4. There was a country \n")


# The custom iterator class
class FileLineReader:
    def __init__(self, filename):
        self.file = open(filename, "r")

    def __iter__(self):
        return self

    def __next__(self):
        line = self.file.readline()
        if line == "":
            self.file.close()
            raise StopIteration
        return line.strip()


# Test it
for line in FileLineReader("sample.txt"):
    print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
from abc import ABC, abstractmethod
import math


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


shapes = [
    Rectangle(12, 7),
    Circle(14),
    Rectangle(6, 8),
    Circle(3),
]

for shape in shapes:
    print(f"Area: {shape.area():.2f}")
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.a > self.n:
            raise StopIteration
        current = self.a
        self.a, self.b = self.b, self.a + self.b
        return current


for num in Fibonacci(51):
    print(num)
````

Create classes `Rectangle`, `Circle`, and `Triangle`, each with an `area()` method.  
- Store instances of these classes in a list.  
- Write code that calculates the total area without checking object types.  

````python
import math


class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"


class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def __str__(self):
        return f"Circle(radius {self.radius})"


class Triangle:
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

    def __str__(self):
        return f"Triangle(base {self.base}, height {self.height})"


shapes = [
    Rectangle(10, 5),
    Circle(14),
    Triangle(8, 6),
    Rectangle(3, 4),
    Circle(7),
    Triangle(9, 3),
]

total_area = 0

for shape in shapes:
    shape_area = shape.area()
    print(f"  {shape} → area: {shape_area:.2f}")
    total_area += shape_area

print(f"\nTotal area: {total_area:.2f}")
````

````python
import math


# ══════════════════════════════════════════════════
# PIECE 1 — Book classes
# ══════════════════════════════════════════════════

class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_borrowed = False

    def get_description(self):
        return f"'{self.title}' by {self.author} ({self.year})"

    def __str__(self):
        status = "Borrowed" if self.is_borrowed else "Available"
        return f"{self.get_description()} [{status}]"


class EBook(Book):
    def __init__(self, title, author, year, file_size, file_format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.file_format = file_format

    def get_description(self):
        return f"[E-Book] '{self.title}' by {self.author} ({self.year}) — {self.file_size}MB, {self.file_format}"


class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return f"[Printed] '{self.title}' by {self.author} ({self.year}) — {self.weight}kg, {self.dimensions}"


# ══════════════════════════════════════════════════
# PIECE 2 — Member classes
# ══════════════════════════════════════════════════

class Member:
    total_borrowed = 0

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = []
        self.borrow_limit = 3

    def borrow_book(self, book):
        if len(self._borrowed_books) >= self.borrow_limit:
            print(f"  ✗ {self.name} has reached their borrowing limit of {self.borrow_limit}!")
            return
        if book.is_borrowed:
            print(f"  ✗ '{book.title}' is already borrowed by someone else.")
            return
        book.is_borrowed = True
        self._borrowed_books.append(book)
        Member.total_borrowed += 1
        print(f"  ✓ {self.name} borrowed '{book.title}'")

    def return_book(self, book):
        if book not in self._borrowed_books:
            print(f"  ✗ {self.name} doesn't have '{book.title}'")
            return
        book.is_borrowed = False
        self._borrowed_books.remove(book)
        print(f"  ✓ {self.name} returned '{book.title}'")

    def get_borrowed_books(self):
        return self._borrowed_books

    def __iter__(self):
        self._index = 0
        return self

    def __next__(self):
        if self._index >= len(self._borrowed_books):
            raise StopIteration
        book = self._borrowed_books[self._index]
        self._index += 1
        return book

    def __str__(self):
        return f"{self.name} (ID: {self.member_id}) — {len(self._borrowed_books)} books borrowed"


class StudentMember(Member):
    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level
        self.borrow_limit = 3

    def borrow_book(self, book):
        print(f"  [Student borrowing — limit: {self.borrow_limit}]")
        super().borrow_book(book)


class TeacherMember(Member):
    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department
        self.borrow_limit = 10

    def borrow_book(self, book):
        print(f"  [Teacher borrowing — limit: {self.borrow_limit}]")
        super().borrow_book(book)


# ══════════════════════════════════════════════════
# PIECE 3 — Library class
# ══════════════════════════════════════════════════

class Library:
    def __init__(self, name):
        self.__library_name = name
        self.books = []
        self.members = []

    def get_library_name(self):
        return self.__library_name

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def get_available_books(self):
        return [book for book in self.books if not book.is_borrowed]

    def get_borrowed_books(self):
        return [book for book in self.books if book.is_borrowed]

    def search_by_title(self, keyword):
        keyword = keyword.lower()
        return [book for book in self.books if keyword in book.title.lower()]

    def search_by_author(self, keyword):
        keyword = keyword.lower()
        return [book for book in self.books if keyword in book.author.lower()]

    def __iter__(self):
        self._index = 0
        return self

    def __next__(self):
        if self._index >= len(self.books):
            raise StopIteration
        book = self.books[self._index]
        self._index += 1
        return book

    def __str__(self):
        return f"Library: {self.__library_name} — {len(self.books)} books, {len(self.members)} members"


# ══════════════════════════════════════════════════
# PIECE 4 — Your code (usage)
# ══════════════════════════════════════════════════

# Create the library
library = Library("Bola Tinubu Learning Center")
print(library)

# Add books
print("\n Adding books...")
book1 = PrintedBook("Things Fall Apart", "Chinua Achebe", 1954, 0.8, "23 x 19 cm")
book2 = PrintedBook("The Accidental Civil Servant", "Nasir El-Rufai", 2000, 0.6, "21 x 15 cm")
book3 = EBook("The Art of War", "Sun Tzu", 1608, 50, "PDF")
book4 = EBook("AI Engineering", "Chip Huyen", 2024, 35, "EPUB")
book5 = PrintedBook("The Pragmatic Programmer", "David Thomas", 2019, 0.7, "23 x 18 cm")

for book in [book1, book2, book3, book4, book5]:
    library.add_book(book)
    print(f"  Added: {book.get_description()}")

# Add members
print("\n Adding members...")
student = StudentMember("Ken Akure", "STU001", "Year 1")
teacher = TeacherMember("Dr. Jubril", "TCH001", "Computer Science")

library.add_member(student)
library.add_member(teacher)
print(f"  {student}")
print(f"  {teacher}")

# Demonstrate borrowing (polymorphism)
print("\n Borrowing books...")
student.borrow_book(book1)
student.borrow_book(book3)
teacher.borrow_book(book2)
teacher.borrow_book(book4)

# Student tries to borrow an already borrowed book
print("\n Trying to borrow an already borrowed book...")
teacher.borrow_book(book1)

# Iterate over library books (iterator)
print("\n All books in library:")
for book in library:
    print(f"  {book}")

# Iterate over a member's borrowed books (iterator)
print(f"\n {student.name}'s borrowed books:")
for book in student:
    print(f"  {book.get_description()}")

# Search books
print("\n Searching for 'war':")
results = library.search_by_title("war")
for book in results:
    print(f"  Found: {book.get_description()}")

print("\n Searching by author 'achebe':")
results = library.search_by_author("achebe")
for book in results:
    print(f"  Found: {book.get_description()}")

# Available vs borrowed
print(f"\n Available books: {len(library.get_available_books())}")
print(f" Borrowed books: {len(library.get_borrowed_books())}")
print(f" Total borrows across all members: {Member.total_borrowed}")

# Return a book
print(f"\n Returning a book...")
student.return_book(book1)

print(f"\n Available books now: {len(library.get_available_books())}")

# Private attribute access
print(f"\n Library name (via getter): {library.get_library_name()}")

# Protected attribute access
print(f" Ken Akure's borrowed books (via getter): {[b.title for b in student.get_borrowed_books()]}")
````
